### Streaming

```python
# inicializar o stemmer do nltk
stemmer = nltk.stem.RSLPStemmer()

doc = nlp(texto_bruto)

for tokens in doc: 
  if token.is_punct or token.is_space:
    continue

  word = token.text
  stem = stemmer.stem(word.lower())
  print(word, "->", stem)
```

### Lematização

```python
# lematização
for tokens in doc: 
  if token.is_punct or token.is_space:
    continue

  word = token.text
  lemma = token.lemma_
  pos = token.pos_
  print(word, "->", lemma, "->", pos)
```

```python
data_comparacao = []

for token in doc:
    if token.is_punct or token.is_space:
        continue

    word = token.text
    stem = stemmer.stem(word.lower())
    lemma = token.lemma_
    pos = token.pos_
    
    # Adicionar a lista
    data_comparacao.append([word, stem, lemma, pos])
data_comparacao
```

### Regex

```python
import re
# funçao para destacar o redultado do RegEX
def mostrar_match(padrao, texto, descricao):
  print(f"--- {descricao} ---")
  print(f"Padrão: {padrao} | Texto: '{texto}'")
  encontrados = re.findall(padrao, texto)
  print(f"Resultados: {encontrados}\n")
  
```

### Metacaracteres, Classes Personalizadas e Quantificadores

```python
texto_exemplo = """O agente 007 ligou às 14h. O código informado foi aeiou_123."""

mostrar_match(r'\d', texto_exemplo, "digitos individuais (\\d)")
```

```python
mostrar_match(r'\w', texto_exemplo, "Alfanuméricos indíviduais (\\w)")
```

```python

mostrar_match(r'\s', texto_exemplo, "Espaços em Branco (\\s)")
```

```python
mostrar_match(r'agente ...', texto_exemplo, "Pont (.) pegando 007")
```

```python
mostrar_match(r'[aeiou]', texto_exemplo, "Apenas as vogais [aeiou]")
```

```python
texto_quant = "a aa aaa ab b aab"
```

```python
mostrar_match(r'a+', texto_quant, "Um ou mais 'a' (a+)")
```

```python
mostrar_match(r'a*b', texto_quant, "Zero ou mais 'a' seguido de 'b' (a*b)")
```

```python
mostrar_match(r'a?b', texto_quant, "Zero ou 1 'a' seguido de 'b' (a?b)")
```

### Grupos de captura e Backreference

```python
data_texto = "Data do evento: 10/01/2026"
padrao_data = r'(\d{2})/(\d{2})/(\d{4})'
```

```python
match = re.search(padrao_data, data_texto)
```

```python
if match:
    print(f'Texto Completo: {match.group(0)}')
    print(f'Dia: {match.group(1)}')
    print(f'Mês: {match.group(2)}')
    print(f'Ano: {match.group(3)}')
```

```python
texto_repetido = "Eu eu vi o o passaro."
```

```python
padrao_repeticao = r'\b(\w+)\s+\1\b'
```

```python
repeticoes = re.findall(padrao_repeticao, texto_repetido, flags=re.IGNORECASE)
print(f'Palavras repetidas: {repeticoes}')
```

### Lookaround (asserções)

```python
texto_precos = "O produto A custa R$ 10,00 e o produto AB custa R$ 20,00"
```

```python
#Extrair apenas o número, sem o R$
padrao_lookbehind = r'(?<=\$\s)\d+,\d{2}'

precos = re.findall(padrao_lookbehind, texto_precos)
print(precos)
```

```python
# Extrair a palavra produto, apenas se for seguida por A
padrao_lookahead = r'produto(?=\sA)'
match_ahead = re.findall(padrao_lookahead, texto_precos)
print(match_ahead)
```

## RegEx em Python
> O módulo **re** oferece funções essenciais para trabalhar com RegEx.

- **re.match(pattern, string):** tenta match no início da string, retorna Match object ou None. 
- **re.search(pattern, string):** busca pattern em qualquer posição da string, retorna primeiro match.
- **re.findall(pattern, string):** retorna lista com todas ocorrências do pattern.
- **re. finditer(pattern, string):** retorna iterador de Match objects para todas ocorrências (mais eficiente que findall para strings grandes).
- **re.sub(pattern, repl, string):** substitui ocorrências do pattern por repl. 
- **re.split(pattern, string):** divide string usando pattern como delimitador.

#### Match objects têm métodos:
- .group retorna string matched,
- .groups() retorna tupla com grupos capturados, .span( retorna (start, end) indices.

```python
texto_lab = "Erro 404: Arquivo não encontrado. Erro 500: Servidor Interno."
```

```python
# Match vs Search
print(re.match(r'Arquivo', texto_lab)) # Retorna no começo da string
print(re.search(r'Arquivo', texto_lab)) # Retorna em qualquer parte da string
```

```python
# Findall vs Findinter
lista_erros = re.findall(r'Erro \d+', texto_lab)
print(lista_erros)
```

```python
['Erro 404', 'Erro 500']
[<re.Match object; span=(0, 8), match='Erro 404'>, <re.Match object; span=(34, 42), match='Erro 500'>]
```

```python
lista_erros_iter = re.finditer(r'Erro (\d+)', texto_lab)
print(list(lista_erros_iter))
```

```python
lista_erros_iter = re.finditer(r'Erro (\d+)', texto_lab)
for match_obj in lista_erros_iter:
    print(match_obj.group())
    print(match_obj.groups()[0])
    print(match_obj.span())
```

```python
novo_texto = re.sub(r'\d{3}', 'XXX', texto_lab)
print(novo_texto)
```

```python
partes = re.split(r':\s', texto_lab)
print(partes)
```
