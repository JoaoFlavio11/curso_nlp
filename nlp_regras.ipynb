```python
# texto com ruido: Pontuação, case variado, Dados sensiveis, formatação pouco inconsistente
texto_bruto = """
    URGENTE: O Cientista de Dados (Senior) enviou um e-mail para rh@empresa.com.br!!!
    O salário proposto é de R$ 12.500,00.
    O CPF do candidato é 123.456.789-00 e ele domina: Python, NLP e REGEX.
    OBS: Nós amamos processamento de linguagem natural...
"""

print(texto_bruto)
```

### Tokenização

```python
# Abosrdagem mais ingênua (ruim para NLP)
tokens_split = texto_bruto.split()
print(tokens_split[:7])
```

```python
# abordagem NLTK
tokens_nltk = nltk.word_tokenize(texto_bruto, language='portuguese')
print(tokens_nltk[:10])
```

```python
# Abordagem com spacy
doc = nlp(texto_bruto)
tokens_spacy = [token.text for token in doc]
print(tokens_spacy[:10])
```

```python
# abordagem nltk com tab
tab_tokenizer = nltk.tokenize.TabTokenizer().tokenize(texto_bruto)
print(tab_tokenizer[:10])
```

```python
#abordagem NLTK por linha
line_tokens = nltk.tokenize.line_tokenize(texto_bruto)
print(line_tokens[:10])
```

```python
# abordagem char tokenizer
char_tokens = [char for char in texto_bruto]
print(char_tokens[:10])
```

### Normalização de texto

```python
# Normalizar o texto por Case
texto_lower = texto_bruto.lower()
print(texto_lower)
```

```python
# Normalizar removendo a pontuação

pontuacao = """!()_-{}[];:.,?<>@#$%ˆ&*˜`'"""
texto_sem_pontuacao = ''.join(char for char in texto_bruto if char not in pontuacao)
print(texto_sem_pontuacao)
print(texto_lower)
```

```python
# remover acentos
import unicodedata
texto_sem_acento = ''.join(char for char in unicodedata.normalize('NFD', texto_bruto) if unicodedata.category(char) != 'Mn')
print(texto_sem_acento)
```

### Streaming

```python
# inicializar o stemmer do nltk
stemmer = nltk.stem.RSLPStemmer()

doc = nlp(texto_bruto)

for tokens in doc: 
  if token.is_punct or token.is_space:
    continue

  word = token.text
  stem = stemmer.stem(word.lower())
  print(word, "->", stem)
```

### Lematização

```python
# lematização
for tokens in doc: 
  if token.is_punct or token.is_space:
    continue

  word = token.text
  lemma = token.lemma_
  pos = token.pos_
  print(word, "->", lemma, "->", pos)
```

```python
data_comparacao = []

for token in doc:
    if token.is_punct or token.is_space:
        continue

    word = token.text
    stem = stemmer.stem(word.lower())
    lemma = token.lemma_
    pos = token.pos_
    
    # Adicionar a lista
    data_comparacao.append([word, stem, lemma, pos])
data_comparacao
```

```python
import pandas as pd
df_comparativo = pd.DataFrame(data_comparacao, columns=["Original", "Stem", "Lemma", "Pos"])

df_filtrado = df_comparativo[df_comparativo['Stem'] != df_comparativo['Lemma']]
df_filtrado.head(20)
```

### Regex

```python
import re
# funçao para destacar o redultado do RegEX
def mostrar_match(padrao, texto, descricao):
  print(f"--- {descricao} ---")
  print(f"Padrão: {padrao} | Texto: '{texto}'")
  encontrados = re.findall(padrao, texto)
  print(f"Resultados: {encontrados}\n")
  
```

### Metacaracteres, Classes Personalizadas e Quantificadores

```python
texto_exemplo = """O agente 007 ligou às 14h. O código informado foi aeiou_123."""

mostrar_match(r'\d', texto_exemplo, "digitos individuais (\\d)")
```

```python
mostrar_match(r'\w', texto_exemplo, "Alfanuméricos indíviduais (\\w)")
```

```python

mostrar_match(r'\s', texto_exemplo, "Espaços em Branco (\\s)")
```

```python
mostrar_match(r'agente ...', texto_exemplo, "Pont (.) pegando 007")
```

```python
mostrar_match(r'[aeiou]', texto_exemplo, "Apenas as vogais [aeiou]")
```

```python
texto_quant = "a aa aaa ab b aab"
```

```python
mostrar_match(r'a+', texto_quant, "Um ou mais 'a' (a+)")
```

```python
mostrar_match(r'a*b', texto_quant, "Zero ou mais 'a' seguido de 'b' (a*b)")
```

```python
mostrar_match(r'a?b', texto_quant, "Zero ou 1 'a' seguido de 'b' (a?b)")
```
